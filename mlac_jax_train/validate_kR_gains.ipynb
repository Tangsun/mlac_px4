```python
import pickle
import numpy as np # Often useful for handling the numerical data within

# Define the path to your saved .pkl file
# This should match the --output_file argument you used when running the tuning script
# For example: 'att_gain_tune_aniso_kr_iso_komega.pkl'
results_filename = 'att_tuning_0601.pkl' # Or whatever you named it

try:
    with open(results_filename, 'rb') as f:
        results_log = pickle.load(f)
    print(f"Successfully loaded results from: {results_filename}")
except FileNotFoundError:
    print(f"ERROR: File not found at {results_filename}")
    exit()
except Exception as e:
    print(f"ERROR: Could not load pickle file. Reason: {e}")
    exit()

# Now, 'results_log' is a list of dictionaries.
# Each dictionary has keys like: 'k_R', 'k_Omega', 'avg_R_error_sq', 'avg_Omega_error_sq', 'combined_error'

# You can inspect the first few entries:
print(f"\nFound {len(results_log)} sets of results.")
if results_log:
    print("\nExample of the first few results entries:")
    for i, entry in enumerate(results_log[:3]): # Print first 3 entries
        print(f"Entry {i}:")
        print(f"  k_R: {entry['k_R']}")
        print(f"  k_Omega: {entry['k_Omega']}")
        print(f"  Avg. R Error Sq: {entry['avg_R_error_sq']:.4e}")
        print(f"  Avg. Omega Error Sq: {entry['avg_Omega_error_sq']:.4e}")
        print(f"  Combined Error: {entry['combined_error']:.4e}")

    # To find the best parameters (lowest combined_error), you can sort the list.
    # Filter out any entries where 'combined_error' might be NaN if simulations failed
    valid_results = [r for r in results_log if not np.isnan(r.get('combined_error', np.nan))]

    if valid_results:
        sorted_results = sorted(valid_results, key=lambda x: x['combined_error'])
        
        print("\nTop 5 best performing gain combinations (lowest combined error):")
        for i, best_entry in enumerate(sorted_results[:5]):
            print(f"Rank {i+1}:")
            print(f"  k_R: {best_entry['k_R']}")
            print(f"  k_Omega: {best_entry['k_Omega']}")
            print(f"  Combined Error: {best_entry['combined_error']:.4e}")
            print(f"  (Avg. R Error Sq: {best_entry['avg_R_error_sq']:.3e}, Avg. Omega Error Sq: {best_entry['avg_Omega_error_sq']:.3e})")

        # You can then take the best k_R and k_Omega for further use:
        if sorted_results:
            best_params = sorted_results[0]
            best_k_R_found = best_params['k_R']
            best_k_Omega_found = best_params['k_Omega']
            print(f"\nOverall best k_R: {best_k_R_found}")
            print(f"Overall best k_Omega: {best_k_Omega_found}")
    else:
        print("\nNo valid results found after filtering NaNs.")

else:
    print("\nNo results were loaded from the file.")
```
